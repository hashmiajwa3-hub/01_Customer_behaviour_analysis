# 3. **Project Overview**
This project provides a comprehensive end-to-end data analysis solution, moving from raw data preparation to actionable business recommendations. The goal is to identify key customer segments, evaluate loyalty patterns, and determine the primary drivers behind purchase behavior.

## 3.1 Objective

The primary goal of this Consumer Behaviour dataset is to better understand its customers shopping behavior in order to improve the sales, customer satisfaction and long-term loyalty with data from 3500+ customers, this dataset empowers us to:

- Identify and profile distinct customer segments based on demographics, purchase frequency, and channel preference
- Determine the relative impact of discounts, product reviews, seasonality, and payment method on purchase decisions and repeat buying
- Compare performance metrics between online and offline channels to identify strengths, gaps, and cross-channel opportunities
- Measure customer loyalty and quantify customer lifetime value by segment
------

## 3.2 Performed Task

Conduct a comprehensive Exploratory Data Analysis (EDA) on the Customer Behaviour Analysis Dataset. 
following a structured workflow:

1. **Data Cleaning** → Handle missing values, duplicates, and inconsistencies
2. **Data Wrangling** → Transform and prepare data for analysis
3. **Exploratory Analysis** → Discover patterns and insights
4. **Business Recommendations** → Summarize findings and actionable insights

## 3.4 Deliverables

### 1. Data Preparation & Modeling (Python)
Located in the `/scripts` folder, the Python notebooks cover:
* Initial data cleaning (handling missing values, outliers, and duplicates).
* Data transformation and feature engineering.
* Exporting the refined dataset for SQL integration.

### 2. Data Analysis (SQL)
The `/sql_queries` directory contains scripts that:
* Model the data into a structured schema.
* Simulate business transactions for deep-dive analysis.
* Extract insights on **Customer Segmentation**, **Loyalty Trends**, and **Purchase Drivers**.

### 3. Visualization & Insights (Power BI)
The interactive dashboard includes:
* KPI Overview (Total Revenue, Customer Count, Retention Rate).
* Trend analysis over time.
* Stakeholder-ready visuals for data-driven decision-making.
