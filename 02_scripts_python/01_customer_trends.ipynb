# Fillna Missing values

We have 2 choces to replace the missing values 
- Median
- Mean


The median is better because mean can be manipulate the data if data has skewness.
As the dataset has different categories so as a beginnear i will not replace the all missing values with whole data median beside that i will calculate the median per category and then will replace the missing values

```python
df['Category'].unique()
```

```python
df["Review Rating"] = pd.to_numeric(df["Review Rating"], errors="coerce")

df_categories = (
    df.groupby("Category")["Review Rating"]
      .agg(["count", "mean", "median", "min", "max"])
)

print(df_categories)
```

```python
# Now replace the missing values in review column with median of each category
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
import pandas as pd

# Assuming 'engine' is your existing connection object
file_path = 'sql.sql'

def save_sql_results(engine, sql_file):
    # 1. Read the long script
    with open(sql_file, 'r') as f:
        # Split by semicolon to handle multiple SELECT statements
        raw_sql = f.read()
        queries = [q.strip() for q in raw_sql.split(';') if q.strip()]

    # 2. Use ExcelWriter to save everything into one file
    with pd.ExcelWriter('sql_results_output.xlsx', engine='openpyxl') as writer:
        for i, query in enumerate(queries):
            try:
                print(f"Executing query {i+1}...")
                # Run the query and load into a DataFrame
                df = pd.read_sql(query, engine)
                
                # Save to a specific sheet (naming it by index or a snippet of the SQL)
                sheet_name = f"Result_Set_{i+1}"
                df.to_excel(writer, sheet_name=sheet_name, index=False)
                
                print(f"  --> Success: {len(df)} rows saved to {sheet_name}")
            except Exception as e:
                print(f"  --> Error on query {i+1}: {e}")

# Call the function using your existing connection
save_sql_results(engine, 'sql.sql')
```
